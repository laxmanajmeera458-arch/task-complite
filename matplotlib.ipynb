```python
a = np.array([22,87,5,43,56,73,55,54,11,20,51,5,79,31,27])
plt.hist(a, bins = [0,25,50,75,100], color='g')
plt.title("histrogram of result")
plt.xticks([0,25,50,75,100])
plt.xlabel('marks')
plt.ylabel('no of students')
plt.show()

import matplotlib.pyplot as plt

marks = [ 55, 62, 67, 70, 90, 92, 95, 98,]

plt.hist(marks, bins=5, color='skyblue')
plt.title('students marks distribution')
plt.xlabel('marks range')
plt.ylabel('number of students')
plt.show()
```

PIE CHART

```python
data = [143378686, 1366417754, 329064917, 270625568, 216565318]

1 = ['china','india','united','indonesia','germany',]

e = [0.1, 0.2, 0.0, 0.0, 0.0]

color = ['#1a4c9c','orange','blue','green','#733238']

plt.pie(data, labels=1, explode=e, shadow=True, startangle = 180,autopct="%1.1f%%",wedgeprops={'edgecolor':'black', 'linewidth': 2}, color=color)

plt.title("population pie chart")
plt.show()
```
